# AMR Project - Step 2: Imputation & Feature Selection (Streamlined)

This notebook covers the missing values check (Step 3) and target/feature alignment (Step 4 & 5) for the Anti-Microbial Resistance (AMR) project.

### Objectives:
1. **Imputation Inspection**: Inspect missingness. Since we kept only the clean biological features (`Taxon ID`, `Genome ID`, `Antibiotic`, `Resistant Phenotype`) during the cleaning phase, **there are 0 missing values to impute**.
2. **Feature Setup**: Confirm the features and target are clean and write the final file.

```python
import os
import pandas as pd
import numpy as np
```

## 1. Load the Cleaned Data

We load the output of the data cleaning notebook from `data/processed/cleaned_data.csv`.

```python
processed_path = "../data/processed/cleaned_data.csv"
print(f"Loading cleaned data from {processed_path}...")
df = pd.read_csv(processed_path, low_memory=False)
print(f"Cleaned dataset shape: {df.shape}")
```

## 2. Verify Missing Values

We verify there are no remaining missing values across any of the selected columns.

```python
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Percentage (%)": (df.isnull().mean() * 100).round(2),
    "Dtype": df.dtypes
})
print(missing_summary)
```

## 3. Save Preprocessed Data

Since the data is completely clean and contains zero missing values, the preprocessed dataset is identical to the cleaned dataset.

```python
output_path = "../data/processed/preprocessed_data.csv"
print(f"Saving preprocessed data to {output_path}...")
df.to_csv(output_path, index=False)
print("Preprocessed data saved successfully!")
```
